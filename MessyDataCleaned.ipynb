```python
import re
def extract_age(age):
    age_num=re.findall('[0-9]+', str(age))
    if len(age_num)>0:
        return age_num[0]
    else:
        return age
df["Age"].apply(lambda x:extract_age(x))
```

```python
df["Age"] = df["Age"].apply(extract_age)
```

```python
df["Age"].unique()
```

```python
df_age=df[df["Age"]!="nan years"]["Age"]
```

```python
df_age
```

```python
age_median=int(df_age.dropna().astype('int64').median())
```

```python
age_median
```

```python
df.replace('nan years',age_median,inplace=True)
```

```python
df['Age'].unique()
```

```python
df['Age'].fillna(age_median,inplace=True)
```

```python
import matplotlib.pyplot as plt 
import seaborn as sns
cols=['Age','Purchase_Amount', 'Feedback_Score' ]
plt.figure(figsize=(15,5))

for i, col in enumerate(cols,1):
    plt.subplot(1,3,i)
    sns.boxplot(y=df[col],color='skyblue',width=0.3)
    plt.title(f'Boxplot of {col}',fontsize=12)
    plt.ylabel('')
    plt.grid(axis='y',linestyle='--',alpha=0.6)

plt.tight_layout()
plt.show()
```

```python
from scipy import stats
import numpy as np
```

```python
z_score=np.abs(stats.zscore(df[['Age','Purchase_Amount']]))
```

```python
df_clean=df[~(z_score>3).any(axis=1)]
```

```python
df_clean.shape
```

```python
import matplotlib.pyplot as plt 
import seaborn as sns
cols=['Age','Purchase_Amount', 'Feedback_Score' ]
plt.figure(figsize=(15,5))

for i, col in enumerate(cols,1):
    plt.subplot(1,3,i)
    sns.boxplot(y=df_clean[col],color='skyblue',width=0.3)
    plt.title(f'Boxplot of {col}',fontsize=12)
    plt.ylabel('')
    plt.grid(axis='y',linestyle='--',alpha=0.6)

plt.tight_layout()
plt.show()
```
